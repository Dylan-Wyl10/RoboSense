```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# path of results files 
coverTable_benchmark = '../result/cover_table_benchmark.npy'
coverTable = '../result/cover_table.npy'
```

```python
# load cover table
cover_benchmark = np.load(coverTable_benchmark)
# cover_benchmark.shape
cover_LinkTime_ben = np.sum(cover_benchmark, axis=2)

# 2%PR
cover = np.load(coverTable)
# cover_benchmark.shape
cover_LinkTime = np.sum(cover, axis=2)
```

```python
# def GetSpacedElements(array, numElems = 4):
#     return out = array[np.round(np.linspace(0, len(array)-1, numElems)).astype(int)]

idx = np.arange(0,4800, 10)
tmp_1 = cover_LinkTime_ben[: ,idx]
tt = np.where(tmp_1 >0, 1, 0)

fig, axs = plt.subplots(
  nrows=2,
  ncols=1,
  sharey=True,
  sharex=True,
  figsize=(15, 12)
)


axs[0].imshow(tmp_1, cmap='viridis', interpolation='none')

axs[1].imshow(tt, cmap='viridis', interpolation='none' )
```

```python
tmp_2 = cover_LinkTime[: ,idx]
tt2 = np.where(tmp_2 > 0, 1, 0)

plt.figure(figsize=(15,12))

fig, axs = plt.subplots(
  nrows=2,
  ncols=1,
  sharey=True,
  sharex=True,
  figsize=(15, 12)
)


axs[0].imshow(tmp_2, cmap='viridis', interpolation='none')

axs[1].imshow(tt2, cmap='viridis', interpolation='none' )
```

```python
aa = np.sum(tt)/tt.size

aaa = np.sum(tt2)/tt2.size

print(aa, aaa)
print(np.sum(tmp_1), np.sum(tmp_2))

tt.size
```
